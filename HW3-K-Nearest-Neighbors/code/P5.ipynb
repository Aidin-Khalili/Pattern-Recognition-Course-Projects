Part a

```python
import os
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.neighbors import DistanceMetric

def load_heart_disease_data(filepath):
    columns = ['age', 'sex', 'cp', 'trestbps', 'chol', 'fbs', 'restecg', 'thalach', 'exang', 'oldpeak', 'slope', 'ca', 'thal', 'target']
    data = pd.read_csv(filepath, names=columns)
    # Converting categorical columns to numerical if necessary
    categorical_columns = ['sex', 'cp', 'fbs', 'restecg', 'exang', 'slope', 'ca', 'thal']
    data = pd.get_dummies(data, columns=categorical_columns)
    # Converting all data to float to avoid conversion issues
    data = data.apply(pd.to_numeric, errors='coerce')
    # Droping rows with NaN values if any
    data = data.dropna()
    return data

# Loading the dataset
filepath = 'F:/Aidin/Universities/Amirkabir (Master)/Lessons/Term 2/Pattern Recognition/Assignments/HW3/Answer/input/P5/heart.csv'
data = load_heart_disease_data(filepath)
print("\n Dataset (first 5 rows):")
print(data.head())
```

Part b

```python
def split_dataset(data):
    train_data, temp_data = train_test_split(data, test_size=0.3, random_state=42)
    validation_data, test_data = train_test_split(temp_data, test_size=0.3333, random_state=42)  # 0.3333 of 0.3 equals about 10%
    return train_data, validation_data, test_data

train_data, validation_data, test_data = split_dataset(data)
print("\nDataset Split:")
print(f"Training Set: {len(train_data)} samples")
print(f"Validation Set: {len(validation_data)} samples")
print(f"Test Set: {len(test_data)} samples")
```

Part c

```python
def evaluate_knn(train_data, validation_data, k_values):
    X_train = train_data.drop('target', axis=1)
    y_train = train_data['target']
    X_validation = validation_data.drop('target', axis=1)
    y_validation = validation_data['target']
    results = {}
    for k in k_values:
        # Initializing k-NN classifier
        knn = KNeighborsClassifier(n_neighbors=k)
        # Training the classifier
        knn.fit(X_train, y_train)
        # Making the predictions on the validation set
        y_pred = knn.predict(X_validation)
        # Calculating metrics
        accuracy = accuracy_score(y_validation, y_pred)
        precision = precision_score(y_validation, y_pred, zero_division=1)
        recall = recall_score(y_validation, y_pred, zero_division=1)
        f1 = f1_score(y_validation, y_pred, zero_division=1)
        # Storing the results
        results[f'{k}-NN'] = {
            'accuracy': accuracy,
            'precision': precision,
            'recall': recall,
            'f1_score': f1
        }
    return results

# Evaluating k-NN with different k values
k_values = [1, 3, 5, 7]
knn_results = evaluate_knn(train_data, validation_data, k_values)

print("\n k-NN Evaluation Results:")
for k, metrics in knn_results.items():
    print(f"\nResults for {k}:")
    for metric, value in metrics.items():
        print(f"{metric.capitalize()}: {value:.2f}")
```

Part d

```python
def evaluate_weighted_knn(train_data, validation_data):
    X_train = train_data.drop('target', axis=1)
    y_train = train_data['target']
    X_validation = validation_data.drop('target', axis=1)
    y_validation = validation_data['target']
    
    # Initialize weighted k-NN classifier (using weights='distance' to give closer neighbors more influence)
    knn_weighted = KNeighborsClassifier(n_neighbors=5, weights='distance')
    # Train the classifier
    knn_weighted.fit(X_train, y_train)
    # Make predictions on the validation set
    y_pred = knn_weighted.predict(X_validation)
    # Calculate metrics
    accuracy = accuracy_score(y_validation, y_pred)
    precision = precision_score(y_validation, y_pred, zero_division=1)
    recall = recall_score(y_validation, y_pred, zero_division=1)
    f1 = f1_score(y_validation, y_pred, zero_division=1)
    # Store the results
    results = {
        'accuracy': accuracy,
        'precision': precision,
        'recall': recall,
        'f1_score': f1
    }
    return results

# Evaluating weighted k-NN
weighted_knn_results = evaluate_weighted_knn(train_data, validation_data)
knn_results['Weighted 5-NN'] = weighted_knn_results
print("\n Weighted k-NN Evaluation Results:")
for metric, value in weighted_knn_results.items():
    print(f"{metric.capitalize()}: {value:.2f}")
```

Part e

```python
import pandas as pd

# Creating a DataFrame to compare results
results_df = pd.DataFrame(knn_results).T
print("\n Comparison of k-NN Models:")
print(results_df)

# Displaying results in a tabular format for better comparison
print(results_df.to_markdown())
```

Part f

```python
best_model_name = 'Weighted 5-NN'
best_model_metrics = results_df.loc[best_model_name]
print(f"\n Best Model Selected: {best_model_name}")
print(best_model_metrics)
```

Part g

```python
def evaluate_best_model_on_test(train_data, test_data, model_name):
    X_train = train_data.drop('target', axis=1)
    y_train = train_data['target']
    X_test = test_data.drop('target', axis=1)
    y_test = test_data['target']
    
    if 'Weighted' in model_name:
        k = int(model_name.split()[1].split('-')[0])
        knn = KNeighborsClassifier(n_neighbors=k, weights='distance')
    else:
        k = int(model_name.split('-')[0])
        knn = KNeighborsClassifier(n_neighbors=k)
    
    knn.fit(X_train, y_train)
    y_pred = knn.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, zero_division=1)
    recall = recall_score(y_test, y_pred, zero_division=1)
    f1 = f1_score(y_test, y_pred, zero_division=1)
    
    print(f"\nTest Set Evaluation for {model_name}")
    print(f"Accuracy: {accuracy:.2f}")
    print(f"Precision: {precision:.2f}")
    print(f"Recall: {recall:.2f}")
    print(f"F1 Score: {f1:.2f}")

evaluate_best_model_on_test(train_data, test_data, best_model_name)
```
